```python
def contains_uppercase(password):
    """Check if password contains at least one uppercase letter."""
    for char in password:
        if 'A' <= char <= 'Z':
            return True
    return False

def contains_lowercase(password):
    """Check if password contains at least one lowercase letter."""
    for char in password:
        if 'a' <= char <= 'z':
            return True
    return False

def contains_digit(password):
    """Check if password contains at least one digit."""
    for char in password:
        if '0' <= char <= '9':
            return True
    return False

def contains_valid_special_characters(password):
    """Check if password contains only valid special characters (!, @, #)."""
    valid_special_chars = "!@#"
    found_special = False
    for char in password:
        if not ('a' <= char <= 'z' or 'A' <= char <= 'Z' or '0' <= char <= '9'):
            if char in valid_special_chars:
                found_special = True
            else:
                return False, True  # Invalid special character found
    return found_special, False  # Valid special character found or not

def validate_password(password, criteria):
    """Validate the password based on selected criteria."""
    if len(password) < 8:
        print(f"'{password}' - Invalid password. Less than 8 Characters.\n")
        return

    errors = []

    if 1 in criteria and not contains_uppercase(password):
        errors.append("Missing Criteria 1: Uppercase letters")

    if 2 in criteria and not contains_lowercase(password):
        errors.append("Missing Criteria 2: Lowercase letters")

    if 3 in criteria and not contains_digit(password):
        errors.append("Missing Criteria 3: Numbers")

    if 4 in criteria:
        has_valid_special, has_invalid_special = contains_valid_special_characters(password)
        if not has_valid_special:
            errors.append("Missing Criteria 4: Special characters (!, @, #)")
        if has_invalid_special:
            errors.append("Contains invalid special characters")

    if errors:
        print(f"'{password}' - Invalid password. " + ", ".join(errors) + "\n")
    else:
        print(f"'{password}' - Valid password.\n")

def main():
    try:
        user_input = input("Enter the criteria numbers to check (e.g., 1,2,3,4): ")
        criteria = {int(x) for x in user_input.split(',') if x.strip().isdigit()}

        if not criteria:
            print("No valid criteria selected. Exiting...")
            return

        password_list = [
            "abc12345",
            "abc",
            "123456789",
            "abcdefg$",
            "abcdefgABHD!@313",
            "abcdefgABHD$$!@313"
        ]

        for password in password_list:
            validate_password(password, criteria)

    except Exception as e:
        print("An error occurred:", e)

if __name__ == "__main__":
    main()
```

```python
def contains_uppercase(password):
    """Check if password contains at least one uppercase letter."""
    for char in password:
        if 'A' <= char <= 'Z':
            return True
    return False

def contains_lowercase(password):
    """Check if password contains at least one lowercase letter."""
    for char in password:
        if 'a' <= char <= 'z':
            return True
    return False

def contains_digit(password):
    """Check if password contains at least one digit."""
    for char in password:
        if '0' <= char <= '9':
            return True
    return False

def contains_valid_special_characters(password):
    """Check if password contains only valid special characters (!, @, #)."""
    valid_special_chars = "!@#"
    found_special = False
    for char in password:
        if not ('a' <= char <= 'z' or 'A' <= char <= 'Z' or '0' <= char <= '9'):
            if char in valid_special_chars:
                found_special = True
            else:
                return False, True  # Invalid special character found
    return found_special, False  # Valid special character found or not

def validate_password(password, criteria):
    """Validate the password based on selected criteria."""
    if len(password) < 8:
        print(f"'{password}' - Invalid password. Less than 8 Characters.\n")
        return False

    errors = []

    if 1 in criteria and not contains_uppercase(password):
        errors.append("Missing Criteria 1: Uppercase letters")

    if 2 in criteria and not contains_lowercase(password):
        errors.append("Missing Criteria 2: Lowercase letters")

    if 3 in criteria and not contains_digit(password):
        errors.append("Missing Criteria 3: Numbers")

    if 4 in criteria:
        has_valid_special, has_invalid_special = contains_valid_special_characters(password)
        if not has_valid_special:
            errors.append("Missing Criteria 4: Special characters (!, @, #)")
        if has_invalid_special:
            errors.append("Contains invalid special characters")

    if errors:
        print(f"'{password}' - Invalid password. " + ", ".join(errors) + "\n")
        return False
    else:
        print(f"'{password}' - Valid password.\n")
        return True

def main():
    try:
        # Get user-selected criteria
        user_input = input("Enter the criteria numbers to check (e.g., 1,2,3,4): ")
        criteria = {int(x) for x in user_input.split(',') if x.strip().isdigit()}

        if not criteria:
            print("No valid criteria selected. Exiting...")
            return

        valid_count = 0
        invalid_count = 0

        # Read passwords from input.txt file
        try:
            with open("/content/drive/MyDrive/input.txt", "r") as file:
                passwords = file.read().splitlines()
        except FileNotFoundError:
            print("Error: input.txt file not found!")
            return

        # Validate passwords
        for password in passwords:
            if validate_password(password, criteria):
                valid_count += 1
            else:
                invalid_count += 1

        # Display results
        print("=" * 50)
        print(f"Total Valid Passwords: {valid_count}")
        print(f"Total Invalid Passwords: {invalid_count}")
        print("=" * 50)

    except Exception as e:
        print("An error occurred:", e)

if __name__ == "__main__":
    main()
```

```python
import re

def validate_password(password, selected_criteria):
    if len(password) < 8:

        return False

    criteria_checks = {
        1: any(char.isupper() for char in password),
        2: any(char.islower() for char in password),
        3: any(char.isdigit() for char in password),
        4: bool(re.search(r'^[^$][!@#][^$]$', password))
    }

    for crit in selected_criteria:
        if not criteria_checks[crit]:

            return False


    return True

def check_passwords_from_list(password_list, selected_criteria):
    valid_count = 0
    invalid_count = 0

    for password in password_list:
        if validate_password(password, selected_criteria):
            valid_count += 1
        else:
            invalid_count += 1

    print(f"Total Valid Passwords: {valid_count}")
    print(f"Total Invalid Passwords: {invalid_count}")

def check_passwords_from_file(filename, selected_criteria):
    try:
        with open(filename, 'r') as file:
            passwords = [line.strip() for line in file.readlines()]
        check_passwords_from_list(passwords, selected_criteria)
    except FileNotFoundError:
        print("Error: input.txt file not found!")

print("Select criteria to enforce (comma-separated numbers):")
print("1 - Uppercase Letters (A-Z)")
print("2 - Lowercase Letters (a-z)")
print("3 - Numbers (0-9)")
print("4 - Special Characters (!, @, #) [Only these 3 allowed]")

selected_criteria = list(map(int, input("Enter criteria numbers: ").split(',')))

password_list = [
    "A737904k*$2004",
    "A1023k",
    "Ankith",
    "abcdefg$",
    "abcdefgABHD!@313",
    "abcdefgABHD$$!@313",
]

print("\nChecking passwords from list:")
check_passwords_from_list(password_list, selected_criteria)

print("\nChecking passwords from input.txt:")
check_passwords_from_file("/content/input.txt", selected_criteria)
```
